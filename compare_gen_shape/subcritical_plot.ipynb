### Plotting the data for area equivalent to $r=\pi/2$

This is using the data with 240 divs

---
### I will begin with subcritical circle and then do the general shape

```julia
using Plots 
using JLD2
sub = jldopen("sub_pi_0708.jld2")
v0 = sub["A"];
v05 = sub["B"];
```

```julia
l=pi
r = pi/2
circ_x = -r:0.01:r
circ_plus = []
circ_minus = []
for i in 1:length(circ_x)
    append!(circ_plus, sqrt(r^2 - circ_x[i]^2))
    append!(circ_minus, -sqrt(r^2 - circ_x[i]^2))
end
xvals = -5l/2:v0[5]:5l/2
pl=heatmap(xvals,xvals,v0[3],
            tickfontsize=16,
            size=(480,400),
            right_margin=0.75Plots.cm)
plot!(pl,circ_x,circ_plus,label=false,color="#00ff22",linestyle=:dot,
        linewidth=3)
plot!(pl,circ_x,circ_minus,label=false,color="#00ff22",linestyle=:dot,  
        linewidth=3)
```

```julia
proj_qr = copy(v0[1])
proj_qi = copy(v0[2])

a = l/2
b = l/2

dx = v0[5]
dy = v0[10]

lhs_x = -5*l/2
lhs_y = -5*l/2


for xi in 1:length(proj_qr[1,:])
	for yi in 1:length(proj_qr[:,1])
		# check if it is in the ellipse
		x = lhs_x + (xi-1)*dx # minus one because we start xi at 1
		y = lhs_y + (yi-1)*dy
		if (x/a)^2 + (y/b)^2 > 1
			proj_qr[yi,xi] = NaN
            proj_qi[yi,xi] = NaN
		end
	end	
end
```

```julia
pl=heatmap(xvals,xvals,proj_qr,
            tickfontsize=16,
            size=(400,400),clims=(-0.5,0.5),colorbar=false,xticks=false,
            yticks=false,xlims=(-2pi/3,2pi/3),ylims=(-2pi/3,2pi/3))
plot!(pl,circ_x,circ_plus,label=false,color="#00ff22",linestyle=:dot,
        linewidth=3)
plot!(pl,circ_x,circ_minus,label=false,color="#00ff22",linestyle=:dot,  
        linewidth=3)
# savefig(pl,"plots/fig4a_q.pdf")
```

```julia
shape_l = sqrt(3)*pi
tic = -5shape_l/2:sv0[5]:5shape_l/2
```

```julia
r1 = sqrt(3)pi/2
circ_x1 = -r1:0.01:r1
circ_plus1 = []
circ_minus1 = []
for i in 1:length(circ_x1)
    append!(circ_plus1, sqrt(r1^2 - circ_x1[i]^2))
    append!(circ_minus1, -sqrt(r1^2 - circ_x1[i]^2))
end
```
