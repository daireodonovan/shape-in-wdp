```julia
using Plots
using JLD2
theme(:vibrant)
```

## Subcritical

```julia
data = jldopen("data/sub_bigger_2506.jld2")
```

```julia
aspect = data["aspects"];
vels = data["vels"];
thrusts = data["thrusts"];
pwrs = data["pwrs"];
effs = data["effs"];
```

```julia
th = contourf(aspect,vels,thrusts, 
        xlabel="Aspect ratio",ylabel="Velocity",
        labelfontsize=15,tickfontsize=13,
        rightmargin=0.5Plots.cm)
```

```julia
ef = contourf(aspect,vels,effs, 
        xlabel="Aspect ratio",ylabel="Velocity",
        labelfontsize=15,tickfontsize=13,
        rightmargin=0.5Plots.cm)
```
