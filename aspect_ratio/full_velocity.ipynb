#### Data prepared from the runfile

```julia
using JLD2
using Plots
using LaTeXStrings
```

```julia
data2 = jldopen("data/full_v_pi_2_div180.jld2")
sub_vels2 = data2["sub_vels"];
sup_vels2 = data2["sup_vels"];
sub_thrust2 = data2["sub_thrust"];
sup_thrust2 = data2["sup_thrust"];
sub_eff2 = data2["sub_eff"];
sup_eff2 = data2["sup_eff"]
```

```julia
y = ones(5)
x = 0.95:0.05:1.15
length(x)

plt = plot(x,y,label=false,fillrange=0,color="grey", fillstyle = :/)
plot!(plt,sub_vels2,sub_thrust2[1,:],
    linewidth=3,
    labelfontsize=16,tickfontsize=14,legendfontsize=16,gridalpha=0.5
    ,ylims=[0,0.3],color="#0072B2",label=L"b/a=0.5",
    size=(800,300))
plot!(plt,sup_vels2[3:20],sup_thrust2[1,3:20],linewidth=3,color="#0072B2",label=false)
plot!(plt,sub_vels2,sub_thrust2[2,:],
    linewidth=3,
    color="#DC267F",label=L"b/a=1")
plot!(plt,sup_vels2[2:20],sup_thrust2[2,2:20],linewidth=3,color="#DC267F",label=false)
plot!(plt,sub_vels2,sub_thrust2[3,:],
    linewidth=3,
    color="#E69F00",label=L"b/a=2")
plot!(plt,sup_vels2,sup_thrust2[3,:],linewidth=3,color="#E69F00",label=false)
# savefig(plt,"fig7a.pdf")
```

```julia
plt2 = plot(x,y,label=false,fillrange=0,color="grey", fillstyle = :/)
plot!(plt2,sub_vels2,sub_eff2[1,:],
    linewidth=3,
    labelfontsize=16,tickfontsize=14,legendfontsize=16,gridalpha=0.5
    ,ylims=[0,0.75],color="#0072B2",label=L"b/a=0.5",
    size=(800,300))
plot!(plt2,sup_vels2[3:20],sup_eff2[1,3:20],linewidth=3,color="#0072B2",label=false)
plot!(plt2,sub_vels2,sub_eff2[2,:],
    linewidth=3,
    color="#DC267F",label=L"b/a=1")
plot!(plt2,sup_vels2[2:20],sup_eff2[2,2:20],linewidth=3,color="#DC267F",label=false)
plot!(plt2,sub_vels2,sub_eff2[3,:],
    linewidth=3,
    color="#E69F00",label=L"b/a=2")
plot!(plt2,sup_vels2,sup_eff2[3,:],linewidth=3,color="#E69F00",label=false,legend=false)
# savefig(plt2,"fig7b.pdf")
```
