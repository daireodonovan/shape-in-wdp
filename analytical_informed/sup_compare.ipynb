### Studying sweep data file 

```julia
# code used to make data 
l = pi
max_rdif = []
max_idif = []
max_dif = []
max_rdif2 = []
max_idif2 = []
max_dif2 = []
thrust_dif = []
thrust_dif2 = []
thrust_ana = []
thrust_num = []
thrust_ana2 = []
op_ana = []
glob_ana = []
op_ana2 = []
glob_ana2 = []
op_num = []
glob_num = []
for vel in 1.1:0.05:2
    A = sup_param(l,l,vel,100,1,false,true)
    # C = sup_param(l,l,vel,100,4,false,true)    
    B = sup_opti(l,l,vel,100,false,true)
    n = length(A[1][:,1])
    real_dif = zeros(n,n)
    imag_dif = zeros(n,n)
    # real_dif2 = zeros(n,n)
    # imag_dif2 = zeros(n,n)
    dif = zeros(n,n)
    # dif2 = zeros(n,n)
    for xi in 1:n
        for yi in 1:n
            re = A[1][yi,xi] - B[1][yi,xi];
            im = A[2][yi,xi] - B[2][yi,xi];
            real_dif[yi,xi] = abs(re);
            imag_dif[yi,xi] = abs(im);
            dif[yi,xi] = sqrt(re^2 + im^2);
            # re2 = C[1][yi,xi] - B[1][yi,xi];
            # im2 = C[2][yi,xi] - B[2][yi,xi];
            # real_dif2[yi,xi] = abs(re2);
            # imag_dif2[yi,xi] = abs(im2);
            # dif2[yi,xi] = sqrt(re2^2 + im2^2);
        end
    end
    append!(max_rdif, norm(real_dif,Inf))
    append!(max_idif, norm(imag_dif,Inf))
    append!(max_dif, norm(dif,Inf))
    # append!(max_rdif2, norm(real_dif2,Inf))
    # append!(max_idif2, norm(imag_dif2,Inf))
    # append!(max_dif2, norm(dif2,Inf))
    # absolute difference in the thrust
    append!(thrust_dif, abs(A[7]-B[7])/B[7])
    append!(thrust_ana, A[7])
    append!(thrust_num, B[7])

    # append!(thrust_dif2, abs(C[7]-B[7])/B[7])
    # append!(thrust_ana2, C[7])


    append!(op_ana, A[15])
    append!(glob_ana, A[16])
    # append!(op_ana2, C[15])
    # append!(glob_ana2, C[16])
    append!(op_num, B[14])
    append!(glob_num, B[15])
    print("done v=",vel,"\n")
end

info = "sup analytically informed comparison ana1: same as subcritical analytical inform, \nana2: kappa>=1 but k<v^2/(v^2-1)"
v = 1.1:0.05:2
# jldsave("data/sup_analytical_compare_1308.jld2"; info,max_rdif,max_rdif2, max_idif,max_idif2,max_dif,max_dif2,thrust_dif,thrust_dif2,thrust_ana,thrust_ana2,thrust_num,op_ana,op_ana2,op_num)
jldsave("data/sup_analytical_compare_1308.jld2"; info,v,max_rdif, max_idif,max_dif,thrust_dif,thrust_ana,thrust_num,op_ana,op_num)
```
